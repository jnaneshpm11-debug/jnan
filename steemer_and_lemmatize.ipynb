```python
article_text='Just what is agility in the context of software enigineering  work? ivar jacobs'
```

```python
import re
import nltk
from nltk.tokenize import word_tokenize
```

```python
article_text = article_text.lower()
article_text
```

```python
clean_text=re.sub(r'[^a-zA-z\s]', '',article_text)
clean_text
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
sentence_list=nltk.sent_tokenize(article_text)
sentence_list
```

```python
tokens=word_tokenize(clean_text)
tokens
```

```python
nltk.download('stopwords')
from nltk.corpus import stopwords
stop_words=set(stopwords.words('english'))
```

```python
word_frequencies={}
for word in tokens:
  if word not in stop_words:
    if word not in word_frequencies:
      word_frequencies[word]=1
    else:
      word_frequencies[word]+=1
word_frequencies
```

```python
from collections import Counter
dictionary=Counter(word_frequencies)
dictionary
```

```python
from nltk.stem import PorterStemmer
from nltk.stem import LancasterStemmer
```

```python
porter=PorterStemmer()
lancaster=LancasterStemmer()

print(porter.stem("cats"))
print(lancaster.stem("cats"))
print(porter.stem("troubling"))
print(lancaster.stem("troubling"))
print(porter.stem("troubled"))
print(lancaster.stem("troubled"))
```

```python
word_list = ["friend", "friendship", "friends", "friendships", "stabil","destabilize"]
print("{:20}{:20}{:20}".format("Word", "Porter Stemmer", "Lancaster Stemmer"))
for word in word_list:
    print("{:20}{:20}{:20}".format(word, porter.stem(word), lancaster.stem(word)))
```

```python
stem_sentence=[]
for word in tokens:
  stem_sentence.append(porter.stem(word))
  stem_sentence.append(" ")
stem_sentence
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
```

```python
import nltk
nltk.download("wordnet")
```

```python
def lemmatize_word(word):
    return lemmatizer.lemmatize(word)
```

```python
word_list=["word1","word2","word3"]
print("{:20}{:20}".format("Word", "Lemma"))
for word in word_list:
  print("{:20}{:20}".format(word, lemmatize_word(word)))
```
